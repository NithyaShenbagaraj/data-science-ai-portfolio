## 2. Explain the results responsibly

- Positive SHAP values push a model output toward the positive class; negative values push it away (for the explained output scale).
- Global importance ranks features by average absolute contribution; it does not show whether a feature is beneficial or harmful by itself.
- Correlated behavioural variables can share or redistribute importance.
- Avoid causal language such as “increasing this feature will make a customer purchase.”
